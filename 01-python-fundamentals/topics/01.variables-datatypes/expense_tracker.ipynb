# Expense Tracker (Basic Version)

A simple command-line script that records an expense using plain variables and keeps a running total, where you can add expenses and subtract discounts.

## What it does

- Asks for a category (for example food or grocery), an amount, and a date, and stores each one in its own variable
- Keeps asking for amounts in a loop and updates a running total
- Adds a normal expense to the total
- Subtracts a discount from the total (for example, a grocery item that was cheaper than the listed price)
- Stops when you enter 0, then prints the final total

## Example

Input:

```
Category: grocery
Date: 2026-10-18
Amount: 500
Amount: 200
Amount (discount, enter as negative): -50
Amount: 0
```
Output:

```
Total for grocery: 650
```

````python
# Input:
category = input("enter the category:")
date = input("enter the date(YYYY-MM-DD):")
total_amount = 0
while True:
    amount = float(input("enter the amount:"))
    if amount == 0:
        break

    if amount < 0:
        print("Amount cannot be negative.")
        continue

    discount = float(input("enter the discount percent :"))
    while discount < 0 or discount > 100 :
        print("Discount must be between 0 and 100.")
        discount = float(input("enter the discount percent :"))

    discount_amount = amount *discount / 100
    final_price = amount - discount_amount
    total_amount += final_price

print("category:" , category)
print("Date:" , date)
print("Total:" , total_amount)
````
